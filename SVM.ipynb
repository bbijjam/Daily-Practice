DATA PREPARATION

```python
import pandas as pd

df = pd.read_excel("Raisin_Dataset.xlsx")
df.sample(5)
```

MODEL EVALUATION: TRAIN TEST SPLIT

```python
X = df[['Area', 'MajorAxisLength','MinorAxisLength', 'Eccentricity','ConvexArea','Extent','Perimeter']]
y= df['Class']
X.head(3)
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.2, random_state = 23)
```

MODEL TRAINING USING SVM: RBF KERNEL :NOT SCALED

```python
from sklearn.svm import SVC

model = SVC(kernel = "rbf")
model.fit(X_train, y_train)

y_pred = model.predict(X_test)

from sklearn.metrics import classification_report

report = classification_report(y_test, y_pred)
print(report)

model.n_iter_
```

MODEL TRAINING USING SVM: LINEAR KERNEL :NOT SCALED

```python
from sklearn.svm import SVC

model = SVC(kernel = "linear")
model.fit(X_train, y_train)

y_pred = model.predict(X_test)

from sklearn.metrics import classification_report

report = classification_report(y_test, y_pred)
print(report)

model.n_iter_
```

MODEL TRAINING USING SVM: POLY KERNEL :NOT SCALED

```python
from sklearn.svm import SVC

model = SVC(kernel = "poly")
model.fit(X_train, y_train)

y_pred = model.predict(X_test)

from sklearn.metrics import classification_report

report = classification_report(y_test, y_pred)
print(report)

model.n_iter_
```

MODEL TRAINING USING SVM: SIGMOID KERNEL :NOT SCALED

```python
from sklearn.svm import SVC

model = SVC(kernel = "sigmoid")
model.fit(X_train, y_train)

y_pred = model.predict(X_test)

from sklearn.metrics import classification_report

report = classification_report(y_test, y_pred)
print(report)

model.n_iter_
```

The different kernels selected drastically varied the models' performance. Choosing the best kernel type is important, and we also need to watch the number of iterations. In the above scenario, the linear type performed way too many iterations than the other kernels but with better performance. 
We can use a Data preprocessing technique called scaling to tackle this issue. 

SCALING

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
scaler.fit(X)

X_train_scaled = scaler.transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

```python
X_train_scaled
```

MODEL TRAINING USING SVM: RBF KERNEL : SCALED

```python
from sklearn.svm import SVC

model = SVC(kernel = "rbf")
model.fit(X_train_scaled, y_train)

y_pred = model.predict(X_test_scaled)

from sklearn.metrics import classification_report

report = classification_report(y_test, y_pred)
print(report)

model.n_iter_
```

MODEL TRAINING USING SVM: Linear KERNEL : SCALED

```python
from sklearn.svm import SVC

model = SVC(kernel = "linear")
model.fit(X_train_scaled, y_train)

y_pred = model.predict(X_test_scaled)

from sklearn.metrics import classification_report

report = classification_report(y_test, y_pred)
print(report)

model.n_iter_
```

MODEL TRAINING USING SVM: POLY KERNEL : SCALED

```python
from sklearn.svm import SVC

model = SVC(kernel = "poly")
model.fit(X_train_scaled, y_train)

y_pred = model.predict(X_test_scaled)

from sklearn.metrics import classification_report

report = classification_report(y_test, y_pred)
print(report)

model.n_iter_
```

MODEL TRAINING USING SVM: SIGMOID KERNEL : SCALED

```python
from sklearn.svm import SVC

model = SVC(kernel = "sigmoid")
model.fit(X_train_scaled, y_train)

y_pred = model.predict(X_test_scaled)

from sklearn.metrics import classification_report

report = classification_report(y_test, y_pred)
print(report)

model.n_iter_
```

As we can see, the number of iterations decreased from millions to thousands after using scaled data. Even the performance of the sigmoid kernel improved.
